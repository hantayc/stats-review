## Quick formula reference

| Formula | Expression |
|---|---|
| Law of Total Probability | $P(A) = \sum_i P(A \mid B_i)\,P(B_i)$ |
| Two-case form | $P(A) = P(A \mid B)P(B) + P(A \mid B^c)P(B^c)$ |
| Bayes' Theorem | $P(B_j \mid A) = \dfrac{P(A \mid B_j)P(B_j)}{\sum_i P(A \mid B_i)P(B_i)}$ |

**Method:** find what you're solving for → pick the cases to split on → write the chance of each case (they add to 1) → write the chance of A in each case → multiply and add → for a reverse question, divide that one case by the total.

Full sheet: `law-of-total-probability-formulas.md`

## Worked example: factory defects

Machines A, B and C make 50%, 30% and 20% of items. Their defect rates are 2%, 3% and 5%.
(a) What's P(defective)? (b) What's P(machine C | defective)?

Here's how you can set it up in code:

```python
priors       = {"A": 0.5,  "B": 0.3,  "C": 0.2}    # P(machine)
p_defect_if  = {"A": 0.02, "B": 0.03, "C": 0.05}   # P(defective | machine)

# (a) Law of Total Probability: weight each case and add
p_defect = sum(p_defect_if[m] * priors[m] for m in priors)
print(f"P(defective) = {p_defect:.4f}")

# (b) Bayes: one branch / total
p_C_given_defect = p_defect_if["C"] * priors["C"] / p_defect
print(f"P(C | defective) = {p_C_given_defect:.4f}")
```

## Problem 2: Commute

It rains 30% of days. When it rains you're late 40% of the time; when it doesn't, you're late 10% of the time.

**(a)** What is P(late)?  
**(b)** Given you were late, what is P(it rained)?

```python
p_rain = 0.3
p_not_rain = 0.7
p_late_given_rain = 0.4
p_late_given_not_rain = 0.1

answer_2a = (p_rain * p_late_given_rain) + (p_not_rain * p_late_given_not_rain)   # P(late)
answer_2b = (p_late_given_rain * p_rain ) / answer_2a    # P(rain | late)

check("2a", answer_2a)
check("2b", answer_2b)
# hint("2a"); hint("2b")
# solution("2a"); solution("2b")
```

## Problem 3: Spam filter

20% of emails are spam. The word "free" appears in 60% of spam emails and 5% of non-spam emails.

**(a)** What is P(email contains "free")?  
**(b)** What is P(spam | contains "free")?

```python
p_spam = 0.2 
p_not_spam = 0.8 
p_free_given_not_spam = 0.05
p_free_given_spam = 0.6 


answer_3a = (p_spam * p_free_given_spam) + (p_not_spam * p_free_given_not_spam)  # P(free)
answer_3b = p_free_given_spam * p_spam / answer_3a   # P(spam | free)

check("3a", answer_3a)
check("3b", answer_3b)
# hint("3a"); hint("3b")
# solution("3a"); solution("3b")
```

## Problem 4: Class sections

Sections 1, 2 and 3 have 20%, 30% and 50% of the students. Their pass rates are 90%, 80% and 70%.

**(a)** What is P(a random student passes)?  
**(b)** Given a student failed, what is P(they're in Section 3)?

```python
import numpy as np 
shares = np.array([0.2, 0.3, 0.5])
pass_rates = np.array([0.9, 0.8, 0.7])

answer_4a = shares @ pass_rates  # P(pass)
answer_4b = ((1 - pass_rates[2]) * shares[2]) / (1-answer_4a)   # P(Section 3 | fail)

check("4a", answer_4a)
check("4b", answer_4b)
# hint("4a"); hint("4b")
# solution("4a"); solution("4b")
```

## Problem 5: Medical test

1% of people have a disease. The test catches 95% of people who have it (sensitivity). It gives a false positive for 5% of healthy people.

**(a)** What is P(positive)?  
**(b)** What is P(disease | positive)? *Before you compute, guess. Is it near 95%?*

```python
p_disease = 0.01 
p_not_disease = 0.99 
p_positive_disease = 0.95 
p_positive_not_disease = 0.05 


my_guess_5b = None   # write your gut guess first!

answer_5a = (p_disease * p_positive_disease) + (p_not_disease * p_positive_not_disease)   # P(+)

# I am noting that this P(disease | +) is precision, of all the positive tests that we had, how many of them actually had the disease.

answer_5b = p_positive_disease * p_disease / answer_5a   # P(disease | +)

check("5a", answer_5a)
check("5b", answer_5b)
# hint("5a"); hint("5b")
# solution("5a"); solution("5b")

# Personal Notes:
# Base rate neglect: recall (sensitivity) is 95%, but precision is only ~16%.
# The test isn't bad; the disease is rare. 5% false positives on the 99% healthy group
# (~49.5 per 1,000) swamps the true positives from the 1% sick group (~9.5 per 1,000).
# Takeaway: P(+ | D) ≠ P(D | +). With a rare positive class, even a low false-positive rate
# drives precision down. Same reason fraud and spam detection is hard.
```
